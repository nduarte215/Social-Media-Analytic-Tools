### LESSON 5: WORKING WITH SPECIFIC **COLUMNS**

```python
# 1. Select one column
if 'Views' in df.columns:
    views_column = df['Views']
    print("1️⃣ Views column (first 5 values):")
    print(views_column.head())
elif len(numeric_columns) > 0:
    # Use first available numeric column
    column_name = numeric_columns[0]
    column_data = df[column_name]
    print(f"1️⃣ {column_name} column (first 5 values):")
    print(column_data.head())

print("\n📝 How to select columns:")
print("   df['column_name'] - gets one column")
print("   df[['col1', 'col2']] - gets multiple columns")
```

```python
# 2. Select multiple columns
if len(df.columns) >= 2:
    first_two_cols = df.columns[:2]
    print(f"\n2️⃣ First two columns ({list(first_two_cols)}):")
    print(df[first_two_cols].head())
```

```python
# 3. Create a new column (basic calculation)
if 'Views' in df.columns and 'Likes' in df.columns:
    df['Engagement_Rate'] = (df['Likes'] / df['Views']) * 100
    print("\n3️⃣ Created new column 'Engagement_Rate':")
    print(df[['Views', 'Likes', 'Engagement_Rate']].head())
elif len(numeric_columns) >= 2:
    col1, col2 = numeric_columns[0], numeric_columns[1]
    df['Ratio'] = df[col1] / df[col2]
    print(f"\n3️⃣ Created ratio of {col1} to {col2}:")
    print(df[[col1, col2, 'Ratio']].head())
```

### LESSON 6: SIMPLE **FILTERING**

```python
# 1. Filter based on a condition
if 'Views' in df.columns:
    high_view_posts = df[df['Views'] > df['Views'].mean()]
    print(f"1️⃣ Posts with above-average views:")
    print(f"   Total posts: {len(df)}")
    print(f"   High-view posts: {len(high_view_posts)}")
    print(f"   Average views: {df['Views'].mean():.1f}")
elif len(numeric_columns) > 0:
    col = numeric_columns[0]
    high_value_rows = df[df[col] > df[col].mean()]
    print(f"1️⃣ Rows with above-average {col}:")
    print(f"   Total rows: {len(df)}")
    print(f"   High-value rows: {len(high_value_rows)}")

print("\n📝 How filtering works:")
print("   df[df['column'] > value] - rows where column > value")
print("   df[df['column'] == 'text'] - rows where column equals 'text'")
```

```python
# 2. Multiple conditions
if 'Views' in df.columns and 'Likes' in df.columns:
    good_posts = df[(df['Views'] > 1) & (df['Likes'] > 1)]
    print(f"\n2️⃣ Posts with Views > 1 AND Likes > 1:")
    print(f"   Found {len(good_posts)} posts")
```

### LESSON 7: BASIC GROUPING AND **AGGREGATION**

```python
# 1. Group by a categorical column (if available)
text_columns = df.select_dtypes(include=['object']).columns
if len(text_columns) > 0:
    group_column = text_columns[0]
    print(f"1️⃣ Grouping by '{group_column}':")

    if len(numeric_columns) > 0:
        grouped = df.groupby(group_column)[numeric_columns[0]].agg(['count', 'mean', 'sum'])
        print(grouped)
    else:
        grouped = df.groupby(group_column).size()
        print(f"Count by {group_column}:")
        print(grouped)

print("\n📝 What is grouping?")
print("   - Like making categories and calculating stats for each")
print("   - Example: Average views by post type")
```

```python
print("\n" + "="*60)
print("🏆 CREATING CHART 3: TOP PERFORMING POSTS")
print("="*60)

# Find the top 10 posts by Views
top_posts = df.nlargest(10, 'Views')

print(f"📊 Found top {len(top_posts)} posts by Views")
print("Top performing posts:")
for i, (idx, post) in enumerate(top_posts.iterrows(), 1):
    print(f"{i}. {post['Views']} views - {post['Post type']} - {post['Publish time'].strftime('%m/%d')}")

# Create the chart
plt.figure(figsize=(14, 8))

# Prepare data for plotting
post_labels = []
views_data = []

for i, (idx, post) in enumerate(top_posts.iterrows()):
     # Create short labels for posts
     date_str = post['Publish time'].strftime('%m/%d')
     post_type = post['Post type'].replace('IG ', '')  # Shorten "IG story" to "story"
     label = f"#{i+1}\n{post_type}\n{date_str}"
     post_labels.append(label)
     views_data.append(post['Views'])

# Create horizontal bar chart (easier to read post labels)
bars = plt.barh(range(len(post_labels)), views_data, color='gold', edgecolor='orange', linewidth=2)

# Customize the chart
plt.title('🏆 Top 10 Posts by Views', fontsize=18, fontweight='bold', pad=20)
plt.xlabel('Views', fontsize=12)
plt.ylabel('Posts', fontsize=12)

# Add value labels on bars
for i, (bar, value) in enumerate(zip(bars, views_data)):
     plt.text(bar.get_width() + max(views_data)*0.01, bar.get_y() + bar.get_height()/2,
            f'{int(value)}', ha='left', va='center', fontweight='bold', fontsize=10)

# Set y-axis labels
plt.yticks(range(len(post_labels)), post_labels, fontsize=9)

# Invert y-axis so #1 post is at the top
plt.gca().invert_yaxis()

# Add grid for easier reading
plt.grid(True, alpha=0.3, axis='x')
plt.tight_layout()
plt.show()

print("✅ CHART 3 COMPLETED!")
print("="*60)
```

```python
print("\n" + "="*60)
print("📈 CREATING CHART 4: ENGAGEMENT COMPARISON")
print("="*60)

# Calculate engagement rate for each post
df['Engagement_Rate'] = ((df['Likes'] + df['Shares'] + df['Replies']) / df['Reach'] * 100).round(2)

# Get top 10 posts by engagement rate
top_engagement_posts = df.nlargest(10, 'Engagement_Rate')

print(f"📊 Top posts by engagement rate:")
for i, (idx, post) in enumerate(top_engagement_posts.iterrows(), 1):
    print(f"{i}. {post['Engagement_Rate']}% - {post['Views']} views - {post['Publish time'].strftime('%m/%d')}")

# Create scatter plot: Views vs Engagement Rate
plt.figure(figsize=(12, 8))

# All posts as light dots
plt.scatter(df['Views'], df['Engagement_Rate'], alpha=0.6, s=60, color='lightblue',
            edgecolors='navy', label='All Posts')

# Highlight top engagement posts
plt.scatter(top_engagement_posts['Views'], top_engagement_posts['Engagement_Rate'],
            s=100, color='red', edgecolors='darkred', label='Top Engagement', zorder=5)

# Add labels for top engagement posts
for idx, post in top_engagement_posts.iterrows():
    plt.annotate(f"{post['Publish time'].strftime('%m/%d')}",
                (post['Views'], post['Engagement_Rate']),
                xytext=(5, 5), textcoords='offset points', fontsize=8,
                bbox=dict(boxstyle='round,pad=0.3', facecolor='yellow', alpha=0.7))

plt.title('📈 Views vs Engagement Rate by Post', fontsize=16, fontweight='bold', pad=20)
plt.xlabel('Views', fontsize=12)
plt.ylabel('Engagement Rate (%)', fontsize=12)
plt.legend()
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("✅ CHART 4 COMPLETED!")
print("="*60)
```
